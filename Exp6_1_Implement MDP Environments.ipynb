```python
import gym
import numpy as np

# ============================================================
# TASK 1: PRINT THE TOTAL NUMBER OF ENVIRONMENTS
# ============================================================
print("========== TASK 1: TOTAL ENVIRONMENTS ==========")
try:
    # Works in older versions of gym
    envs = gym.envs.registry.all()
    total_envs = len(envs)
except AttributeError:
    # Works in modern gym / gymnasium
    envs = gym.envs.registry.keys()
    total_envs = len(envs)

print(f"Total number of environments: {total_envs}\n")
```

```python
# ============================================================
# TASK 2: PRINT ALL ENVIRONMENTS
# ============================================================
print("========== TASK 2: ALL ENVIRONMENT NAMES ==========")

try:
    # Works in older versions of gym
    env_names = sorted([env_spec.id for env_spec in gym.envs.registry.all()])
except AttributeError:
    # Works in modern gym / gymnasium
    env_names = sorted(list(gym.envs.registry.keys()))

# Printing the first 20 as a sample to avoid flooding the terminal
print(f"Printing first 20 out of {len(env_names)} environments:")
for name in env_names[:20]:
    print(f" - {name}")
print("... (truncated for readability) ...\n")

```

```python
# ============================================================
# TASK 3: EXPLORE RL ENVIRONMENTS AND DEFINE THEIR MDPs
# ============================================================
print("========== TASK 3: ENVIRONMENT EXPLORATION & MDPs ==========\n")

target_envs = {
    "CartPole-v1": "CartPole (Inverted Pendulum balancing)",
    "FrozenLake-v1": "Frozen Lake (Gridworld navigation on ice)",
    "MountainCar-v0": "Mountain Car (Acquiring momentum to climb a hill)",
    "Blackjack-v1": "Blackjack (Card game strategy)",
    "Taxi-v3": "Taxi (Passenger pickup and drop-off navigation)",
    "CliffWalking-v0": "Cliff Walking (Grid navigation avoiding grid pitfalls)"
}

for env_id, description in target_envs.items():
    print('=' * 60)
    print(f"ENVIRONMENT: {env_id} ({description})")
    print('=' * 60)

    try:
        # Initialize environment
        env = gym.make(env_id)

        # 1. Display Spaces
        print(f"🤖 Action Space:      {env.action_space}")
        print(f"👁️  Observation Space: {env.observation_space}")

        # 2. Programmatic MDP Component Analysis
        print("\n📝 MDP Components:")

        # State Space Detail
        if isinstance(env.observation_space, gym.spaces.Discrete):
            print(f"   • States (S): Discrete space with {env.observation_space.n} unique states.")
        elif isinstance(env.observation_space, gym.spaces.Box):
            print(f"   • States (S): Continuous space. Shape: {env.observation_space.shape}")
            print(f"     Low bounds:  {env.observation_space.low}")
            print(f"     High bounds: {env.observation_space.high}")
        elif isinstance(env.observation_space, gym.spaces.Tuple):
            print(f"   • States (S): Tuple space containing: {env.observation_space.spaces}")

        # Action Space Detail
        if isinstance(env.action_space, gym.spaces.Discrete):
            print(f"   • Actions (A): Discrete space with {env.action_space.n} valid actions.")
        elif isinstance(env.action_space, gym.spaces.Box):
            print(f"   • Actions (A): Continuous space with shape {env.action_space.shape}")

        # Transitions & Rewards Detail (Inspecting internal P dictionary for discrete envs)
        if hasattr(env, 'P') or hasattr(env.unwrapped, 'P'):
            P_dict = env.P if hasattr(env, 'P') else env.unwrapped.P
            print("   • Transitions (P) & Rewards (R): Stochastic/Discrete Transition Table found.")
            # Sample a single entry from the transition matrix if available
            sample_state = 0
            if sample_state in P_dict:
                sample_action = 0
                print(f"     Sample transition from State {sample_state} taking Action {sample_action}:")
                # Format: (transition_probability, next_state, reward, done)
                for prob, next_state, reward, done in P_dict[sample_state][sample_action]:
                    print(f"       -> Prob: {prob}, Next State: {next_state}, Reward: {reward}, Done: {done}")
        else:
            print("   • Transitions (P) & Rewards (R): Analytical/Physics equations handle transitions step-by-step.")
            print("     Rewards are computed dynamically via code formulas at each transition boundary.")

        env.close()
        print("\n")

    except Exception as e:
        print(f"Could not load {env_id}: {e}\n")
```
